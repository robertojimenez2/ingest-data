```python
import pandas as pd
```

```python
# read csv
df = pd.read_csv("atp_tennis.csv", low_memory=False)
```

```python
df.dtypes
```

```python
df.head()
```

```python
dtype = {
    "tournament": "string",
    "date": "string",
    "series": "string",
    "court": "string",
    "surface": "string",
    "round": "string",
    "BestOf": "Int64",
    "player1": "string",
    "player2": "string",
    "winner": "string",
    "rank1": "Int64",
    "rank2": "Int64",
    "pts1": "Int64",
    "pts2": "Int64",
    "odds1": "Float64",
    "odds2": "Float64",
    "score": "string"
}

parse_dates = [
    "Date"
    ]

```

```python
df['Date'] = pd.to_datetime(df['Date'])
df['Odd_1'] = pd.to_numeric(df['Odd_1'], errors='coerce')
df['Odd_2'] = pd.to_numeric(df['Odd_2'], errors='coerce')
```

```python
# clean df
df = df[(df['Date'] >= '2010-01-01') & (df['Odd_1'] > 0) & (df['Odd_2'] > 0)].copy()
df = df.sort_values('Date').reset_index(drop=True)
```

```python
df['target'] = (df['Player_1'] == df['Winner']).astype(int)
```

```python
df.head()
```

```python
from sqlalchemy import create_engine
engine = create_engine('postgresql+psycopg://root:root@localhost:5432/my_taxi')
```

```python
print(pd.io.sql.get_schema(df, name='tenis_data', con=engine))
```

```python
df.head(n=0).to_sql(name='tenis_data', con=engine, if_exists='replace')
```

```python
df_iter = pd.read_csv(
    'atp_tennis.csv',
    dtype=dtype,
    parse_dates=parse_dates,
    iterator=True,
    chunksize=100000
)
```

```python
for df_chunk in df_iter:
    print(len(df_chunk))
```

```python
df_chunk.to_sql(name='tenis_data', con=engine, if_exists='append')
```

```python
from tqdm.auto import tqdm

first_chunk = True

for df_chunk in tqdm(df_iter):
    # 2. Filtrado eficiente (Suponiendo formato YYYY-MM-DD en el CSV)
    # Es más rápido filtrar como string antes de convertir a datetime
    df_filtered = df_chunk[df_chunk['Date'] >= '2010-01-01'].copy()
    
    if df_filtered.empty:
        continue

    # Opcional: Convertir a datetime solo los datos que sí se van a insertar
    df_filtered['Date'] = pd.to_datetime(df_filtered['Date'])

    # 3. Inserción optimizada
    if first_chunk:
        # Crea la tabla (replace) con el primer bloque de datos reales
        df_filtered.to_sql(
            name='tennis_data', 
            con=engine, 
            if_exists='replace', 
            index=False,
            method='multi', # <--- CLAVE PARA VELOCIDAD
            chunksize=1000  # Tamaño del bloque de inserción SQL
        )
        first_chunk = False
    else:
        # Añade el resto (append)
        df_filtered.to_sql(
            name='tenis_data', 
            con=engine, 
            if_exists='append', 
            index=False,
            method='multi',
            chunksize=1000
        )
```
